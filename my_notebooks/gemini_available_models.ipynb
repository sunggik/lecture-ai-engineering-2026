```python
import json
import os
from urllib.error import HTTPError, URLError
from urllib.request import Request, urlopen

api_key = os.environ.get("GEMINI_API_KEY_FOR_AIE_2026")
if not api_key:
    raise RuntimeError("環境変数 GEMINI_API_KEY_FOR_AIE_2026 が設定されていません。")

url = "https://generativelanguage.googleapis.com/v1beta/models?pageSize=1000"
models = []

while url:
    request = Request(url, headers={"x-goog-api-key": api_key})

    try:
        with urlopen(request, timeout=30) as response:
            data = json.load(response)
    except (HTTPError, URLError) as error:
        raise RuntimeError(f"Gemini APIへのアクセスに失敗しました: {error}") from error

    for model in data.get("models", []):
        methods = model.get("supportedGenerationMethods", [])
        if "generateContent" in methods:
            models.append({
                "name": model.get("name"),
                "displayName": model.get("displayName"),
                "supportedGenerationMethods": methods,
            })

    next_token = data.get("nextPageToken")
    url = (
        "https://generativelanguage.googleapis.com/v1beta/models"
        f"?pageSize=1000&pageToken={next_token}"
        if next_token else None
    )

print("無料枠の対象かどうかは、このAPI応答からは判定できません。")
print(json.dumps(models, ensure_ascii=False, indent=2))
```

```python
for model in models:
    model_name = model['name']
    if 'gemini' in model_name.lower():
        print(f"モデル名: {model_name}")
```
